```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
# Load the dataset
data=pd.read_csv("student_performance_prediction.csv")
#chek colum name and data type
print(data.info())
# Check missing values
print (data.isnull().sum())

# Check duplicates without Student ID
duplicate_count = data.duplicated(
    subset=data.columns.drop("Student ID")
).sum()

print("Duplicates without Student ID:", duplicate_count)

# Remove duplicates before splitting
data = data.drop_duplicates(
    subset=data.columns.drop("Student ID")
).copy()

print(
    "Duplicates remaining:",
    data.duplicated(subset=data.columns.drop("Student ID")).sum()
)

# Remove rows where the target is missing
data=data.dropna(subset = ["Passed"]).copy()
# Separate features and target
x = data.drop(columns=["Passed", "Student ID"])
y=data["Passed"]
# Split the data
x_train, x_test , y_train , y_test= train_test_split(
 x,y , test_size=0.2 , random_state=42
 )
print("Duplication:" , data.duplicated().sum())
data = data.drop_duplicates().copy()
#idenfie numerical ,categorical columns
numerical_columns=x_train.select_dtypes(
    include="number"
    ).columns.tolist()
categorical_columns=x_train.select_dtypes(
    include="object"
    ).columns.tolist()
#fill numerical missing value using training,madian
for col in numerical_columns:
    median_values=x_train[col].median()
    x_train[col]=x_train[col].fillna(median_values)
    x_test[col]=x_test[col].fillna(median_values)
#fill categorical missing value using training mode

for col in categorical_columns:
    mode_values=x_train[col].mode()[0]
    x_train[col]=x_train[col].fillna(mode_values)
    x_test[col]= x_test[col].fillna(mode_values)
#check missing value
print("training in missing:",x_train.isnull().sum().sum())
print("testing in missing:", x_test.isnull().sum().sum())
print("training shape:" , x_train.shape)
print("test in shape:" , x_test.shape)

#Encoding categorcail columns
x_train=pd.get_dummies(
    x_train,
    columns=categorical_columns,
    dtype=int
    )
x_test=pd.get_dummies(
    x_test,
    columns=categorical_columns,
    dtype=int
    )
#make sure both the dataset have the sane columns

x_test=x_test.reindex(
   columns=x_train.columns ,
    fill_value=0

    )

print("training columns:", x_train.columns.tolist())
print("shape testing:",x_test.shape)
print("\nFirst five rows:")
print(x_train.head())

#standardize
numeric_columns = [
    "Study Hours per Week",
    "Attendance Rate",
    "Previous Grades"
]


# Create scaler
scaler = StandardScaler()

# Fit and transform training data
x_train[numerical_columns] = scaler.fit_transform(
    x_train[numerical_columns]
)

# Transform testing data using the same scaler
x_test[numerical_columns] = scaler.transform(
    x_test[numerical_columns]
)

# Check training means
print("Training means:")
print(x_train[numerical_columns].mean())

# Check training standard deviations
print("Training standard deviations:")
print(x_train[numerical_columns].std())

print("Training data after standardization:")
print(x_train[numerical_columns].head())

y_train = y_train.map({"No": 0, "Yes": 1})
y_test = y_test.map({"No": 0, "Yes": 1})
```
